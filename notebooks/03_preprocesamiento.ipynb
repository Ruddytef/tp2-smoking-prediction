# Preprocesamiento

#### Dependencias

```python
import pandas as pd
import numpy as np
import joblib
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import MinMaxScaler
```

#### Parámetros

```python
SEED = 42
```

#### Lectura

```python
df = pd.read_csv("../data/processed/smoking_train_corregido.csv")
df.shape
```

#### Columnas que no aportan

```python
df = df.drop(columns=["ID", "oral"]) # ID es un identificador y oral tiene un único valor
df.shape # deberían quedar 25 columnas
```

#### Separación de variables

```python
X = df.drop(columns=["smoking"]) # las variables que usa el modelo para predecir
y = df["smoking"]                # lo que queremos predecir
```

```python
X.shape, y.shape
```

#### Train / test

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=SEED, stratify=y)
```

```python
y_train.value_counts(normalize=True) * 100 # % de fumadores en entrenamiento
```

```python
y_test.value_counts(normalize=True) * 100 # % de fumadores en prueba
```

Separé 70% para entrenar y 30% para evaluar. Con stratify, los dos grupos mantienen la misma proporción de fumadores (36,7%).

#### Valores extremos

```python
columnas_recortar = ["eyesight(left)", "eyesight(right)", "systolic", "relaxation",
                     "fasting blood sugar", "Cholesterol", "triglyceride", "HDL", "LDL",
                     "AST", "ALT", "Gtp", "serum creatinine"]
```

```python
limites = X_train[columnas_recortar].quantile(0.99) # el percentil 99 de cada columna, calculado SOLO con train
limites
```

```python
X_train[columnas_recortar] = X_train[columnas_recortar].clip(upper=limites, axis=1)
X_test[columnas_recortar] = X_test[columnas_recortar].clip(upper=limites, axis=1)
```

```python
X_train[["LDL", "ALT", "eyesight(left)"]].max() # ahora los máximos ya no son absurdos
```

Recorté los valores extremos de los análisis de sangre, la presión y la vista al percentil 99, calculado solo con train. No usé la media porque una persona con un valor muy alto quedaría como alguien promedio. Así se corrigen el LDL de 1.860, la vista de 9,9 y los 999, sin borrar filas.

#### Codificación (OHE)

```python
cat_cols = ["gender", "tartar"] # las columnas de texto que quedan
```

```python
ohe = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
ohe.fit(X_train[cat_cols]) # el fit SIEMPRE con train
```

```python
X_train_ohe = pd.DataFrame(ohe.transform(X_train[cat_cols]), columns=ohe.get_feature_names_out(cat_cols), index=X_train.index)
X_test_ohe = pd.DataFrame(ohe.transform(X_test[cat_cols]), columns=ohe.get_feature_names_out(cat_cols), index=X_test.index)
X_train_ohe.head()
```

```python
X_train_final = pd.concat([X_train.drop(columns=cat_cols), X_train_ohe], axis=1)
X_test_final = pd.concat([X_test.drop(columns=cat_cols), X_test_ohe], axis=1)
X_train_final.shape, X_test_final.shape
```

Codifiqué gender y tartar con One Hot Encoding. El encoder aprendió las categorías solo con train y lo apliqué igual a test.

#### Scaler (para KNN)

```python
scaler = MinMaxScaler()
scaler.fit(X_train_final) # el fit SIEMPRE con train
```

El scaler solo lo uso para KNN, porque compara distancias entre personas y necesita que todas las variables estén en la misma escala. Los modelos de árboles no lo necesitan.

#### Guardar

```python
joblib.dump(limites, "../models/limites_outliers.joblib")
joblib.dump(ohe, "../models/ohe.joblib")
joblib.dump(scaler, "../models/scaler.joblib")
```

```python
X_train_final.to_csv("../data/processed/X_train.csv", index=False)
X_test_final.to_csv("../data/processed/X_test.csv", index=False)
y_train.to_csv("../data/processed/y_train.csv", index=False)
y_test.to_csv("../data/processed/y_test.csv", index=False)
```

Guardé el encoder, el scaler y los límites de los outliers en la carpeta models, para aplicarlos igual en la predicción final. Los datos de train y test quedaron en data/processed, listos para entrenar.
