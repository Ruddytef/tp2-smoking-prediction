Las columnas age, height y weight tienen valores lógicos, pero las 18 columnas con decimales no: la cintura va de 2 a 5 cm y la presión sistólica no pasa de 10. Al investigar, descubrí que al pasar por Google Sheets estos números se guardaron como horas del reloj. Por ejemplo, la presión 114 quedó como 114 horas, que Excel guarda como 114 ÷ 24 = 4,75 días. Los decimales se leyeron como minutos: una cintura de 81,3 cm se guardó como "81 horas y 3 minutos". Se pueden recuperar los valores reales multiplicando por 24 y convirtiendo los minutos de vuelta a decimales.

El promedio de smoking es 0,367: el 36,7% de las personas fuma.

El archivo de predicción tiene 5.692 filas y 26 columnas, sin valores nulos. Tiene las mismas columnas que el de entrenamiento, menos smoking, que es lo que el modelo debe predecir. También tiene la columna oral con un solo valor y el mismo problema de los números guardados como horas, así que necesita exactamente las mismas correcciones.

#### Funciones

```python
def corregir_horas(df, columnas):
    for col in columnas:
        horas = df[col] * 24                          # pasa el valor a horas
        enteras = np.floor(horas + 0.000001)          # parte entera: las horas
        minutos = np.round((horas - enteras) * 60)    # lo que sobra, pasado a minutos
        df[col] = enteras + minutos / 10              # los minutos eran el decimal original
    return df
```

#### Corrección de las columnas

```python
columnas_horas = df.select_dtypes(include="float").columns.tolist()
columnas_horas # las 18 columnas con decimales afectadas
```

```python
df = corregir_horas(df, columnas_horas)
df_pred = corregir_horas(df_pred, columnas_horas)
```

```python
df[["waist(cm)", "systolic", "hemoglobin", "eyesight(left)"]].describe() # ahora con valores reales
```

Después de corregir, los valores ya son lógicos: cintura entre 51 y 129 cm, presión entre 71 y 240, hemoglobina entre 4,9 y 21,1. En eyesight aparece un máximo de 9,9, que no parece una medida real; lo reviso en el EDA.
